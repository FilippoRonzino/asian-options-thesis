```python
# Load the Excel file
xls = pd.ExcelFile("/Users/filipporonzino/Downloads/prices.xlsx")
df_full = pd.read_excel(xls, sheet_name="Sheet1", header=None)

# Identify columns for tickers and prices
ticker_columns = list(range(2, df_full.shape[1], 3))  # C5, F5, I5, ...
price_columns = list(range(3, df_full.shape[1], 3))  # D6, G6, J6, ...

# Extract tickers and clean names (remove "Comdty" and extra spaces)
tickers = df_full.iloc[4, ticker_columns].astype(str).tolist()

# Extract date column (C7 downwards)
date_column = df_full.iloc[6:, 2].reset_index(drop=True)  # Column C (index 2)

# Extract price data (D7 downwards)
prices = df_full.iloc[6:, price_columns].reset_index(drop=True)

# Combine into DataFrame
df_final = pd.concat([date_column, prices], axis=1)

# Assign column names
df_final.columns = ["Date"] + tickers

# Convert Date column to datetime
df_final["Date"] = pd.to_datetime(df_final["Date"], errors="coerce")

# Convert price columns to numeric
df_final.iloc[:, 1:] = df_final.iloc[:, 1:].apply(pd.to_numeric, errors="coerce")

# Display cleaned DataFrame
df_final
```

```python
df_final = df_final[(df_final["Date"] >= df_final["Date"].min()) & (df_final["Date"] <= df_final["Date"].max())]
```

```python
df_final
```

```python
# keep just columns with at least 50 percent of non-NaN values
df_final = df_final.dropna(axis=1, thresh=0.5*len(df_final))
```
