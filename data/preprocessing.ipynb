```python
# Load the Excel file
xls = pd.ExcelFile("/Users/filipporonzino/Downloads/prices.xlsx")
df_full = pd.read_excel(xls, sheet_name="Sheet1", header=None)

# Identify columns for tickers and prices
ticker_columns = list(range(2, df_full.shape[1], 3))  # C5, F5, I5, ...
price_columns = list(range(3, df_full.shape[1], 3))  # D6, G6, J6, ...

# Extract tickers and clean names (remove "Comdty" and extra spaces)
tickers = df_full.iloc[4, ticker_columns].astype(str).tolist()

# Extract date column (C7 downwards)
date_column = df_full.iloc[6:, 2].reset_index(drop=True)  # Column C (index 2)

# Extract price data (D7 downwards)
prices = df_full.iloc[6:, price_columns].reset_index(drop=True)

# Combine into DataFrame
df_final = pd.concat([date_column, prices], axis=1)

# Assign column names
df_final.columns = ["Date"] + tickers

# Convert Date column to datetime
df_final["Date"] = pd.to_datetime(df_final["Date"], errors="coerce")

# Convert price columns to numeric
df_final.iloc[:, 1:] = df_final.iloc[:, 1:].apply(pd.to_numeric, errors="coerce")

# Display cleaned DataFrame
df_final
```

```python
df_final = df_final[(df_final["Date"] >= df_final["Date"].min()) & (df_final["Date"] <= df_final["Date"].max())]
```

```python
df_final
```

```python
# keep just columns with at least 50 percent of non-NaN values
df_final = df_final.dropna(axis=1, thresh=0.5*len(df_final))
```

```python
df_final 
df_final.columns = df_final.columns.str.strip().str.replace(r'\s+', ' ', regex=True)
print(df_final.columns)
```

```python
universe = universe.drop(columns=["id().ORIG_IDS:0", "id().ORIG_IDS:1", "DATES"])
universe = universe.rename(columns={"id().ORIG_IDS:0": "ID", "security_des()": "SECURITY_DES", "undl_ticker().EXERCISE_TYPE": "EXERCISE_TYPE", "undl_ticker().EXPIRE_DT": "EXPIRE_DT", "undl_ticker().PUT_CALL": "PUT_CALL", "undl_ticker().STRIKE_PX": "STRIKE_PX", "undl_ticker()": "TICKER"})
universe
```

```python
universe["SECURITY_DES"] = universe["SECURITY_DES"].str.strip().str.replace(r'\s+', ' ', regex=True)
universe
```

```python
xls = pd.ExcelFile("/Users/filipporonzino/Downloads/voldata.xlsx")
df_vol = pd.read_excel(xls, sheet_name="Sheet1", header=5)
df_vol
```

```python


# Step 1: Extract dates from df_final
dates = df_final[["Date"]].drop_duplicates()

# Step 2: Create cross join (assign key, merge, then drop key)
universe["key"] = 1
dates["key"] = 1
universe_expanded = universe.merge(dates, on="key").drop(columns=["key"])

# Step 3: Melt df_final to long format (reshape so SECURITY_DES is a column)
df_long = df_final.melt(id_vars=["Date"], var_name="SECURITY_DES", value_name="OPT_PX")

# Step 4: Merge expanded universe with price data
universe_expanded = universe_expanded.merge(df_long, on=["SECURITY_DES", "Date"], how="left")

# Print the first few rows
universe_expanded
```

```python
# Step 5: Merge universe_expanded with df_vol on Date
universe_expanded = universe_expanded.merge(df_vol, left_on="Date", right_on="Dates", how="left").drop(columns=["Dates"])

universe_expanded
```

```python
# use as risk free rate the U.S. 10 Year Treasury Yield from Yahoo Finance from min date to max date
import yfinance as yf

risk_free = yf.download("^TNX", start=universe_expanded["Date"].min(), end=universe_expanded["Date"].max())
risk_free
```

```python
# add the risk free rate to the universe_expanded DataFrame
universe_expanded = universe_expanded.merge(risk_free["Close"], left_on="Date", right_index=True, how="left").rename(columns={"Close": "Risk_Free_Rate"})
```

```python
universe_expanded
```

```python
# drop nan
universe_expanded = universe_expanded.dropna()
# rename ^TNX column to Risk_Free_Rate
universe_expanded = universe_expanded.rename(columns={"^TNX": "RISK_FREE_RATE"})
universe_expanded["RISK_FREE_RATE"] = universe_expanded["RISK_FREE_RATE"] / 100
universe_expanded
```

```python
# from Yahoo Finance download the WTI Crude Oil price from min date to max date
wti = yf.download("CL=F", start=universe_expanded["Date"].min(), end=universe_expanded["Date"].max()+pd.Timedelta(days=1))
wti
```

```python
# add the WTI Crude Oil price to the universe_expanded DataFrame
universe_expanded = universe_expanded.merge(wti["Close"], left_on="Date", right_index=True, how="left")
universe_expanded = universe_expanded.rename(columns={"CL=F": "UNDERLYING_PRICE"})
```

```python
universe_expanded
```

```python
# annualize the EXPIRE_DT to calculate the time to maturity
universe_expanded["EXPIRE_DT"] = pd.to_datetime(universe_expanded["EXPIRE_DT"])
universe_expanded["TIME_TO_MATURITY"] = (universe_expanded["EXPIRE_DT"] - universe_expanded["Date"]).dt.days / 365

universe_expanded
```
